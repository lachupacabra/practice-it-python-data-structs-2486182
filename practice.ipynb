```python
import time
from collections import deque

# List - slow at front removal
lst = list(range(100000))
start = time.time()
for _ in range(1000):
    lst.pop(0)
print(f"List: {time.time() - start:.4f}s")

# Deque - fast at front removal
dq = deque(range(100000))
start = time.time()
for _ in range(1000):
    dq.popleft()
print(f"Deque: {time.time() - start:.4f}s")
# Deque is much faster!
```

```python
from collections import namedtuple

# Create a named tuple type
Point = namedtuple('Point', ['x', 'y'])

# Create an instance
p = Point(10, 20)
print(p)           # Point(x=10, y=20)
print(p.x)         # 10
print(p[0])        # 10 (still a tuple!)
```
